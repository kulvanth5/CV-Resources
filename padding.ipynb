```python
from google.colab import files
uploaded=files.upload()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
img = cv2.imread(list(uploaded.keys())[0], cv2.IMREAD_GRAYSCALE)

print(img.shape)
plt.imshow(img, cmap='gray')
plt.title("original Image")
plt.axis('off')
```

```python
# Zero Padding
zero_pad = cv2.copyMakeBorder(img, 50, 50, 50, 50,
                              cv2.BORDER_CONSTANT,
                              value=0)
plt.figure(figsize=(10,5))

plt.subplot(1,2,1)
plt.title("Original Image")
plt.imshow(img, cmap='gray')
plt.axis('off')

plt.subplot(1,2,2)
plt.title("Zero Padding")
plt.imshow(zero_pad, cmap='gray')
plt.axis('off')

print("Original Shape:", img.shape)
print("Zero Padding Shape:", zero_pad.shape)

plt.show()
```

```python
# Replicate Padding
reflect_pad = cv2.copyMakeBorder(img, 100, 100, 100, 100,
                                 cv2.BORDER_REPLICATE
                              )
plt.figure(figsize=(10,5))

plt.subplot(1,2,1)
plt.title("Original Image")
plt.imshow(img, cmap='gray')
plt.axis('off')

plt.subplot(1,2,2)
plt.title("Reflect Padding ")
plt.imshow(reflect_pad, cmap='gray')
plt.axis('off')

print("Original Shape:", img.shape)
print("Reflect Padding Shape:", reflect_pad.shape)

plt.show()
```

```python
# Mirror / Symmetric Padding
mirror_pad = cv2.copyMakeBorder(img, 100, 100, 100, 100,
                                cv2.BORDER_REFLECT_101)

plt.figure(figsize=(10,5))

plt.subplot(1,2,1)
plt.title("Original Image")
plt.imshow(img, cmap='gray')
plt.axis('off')

plt.subplot(1,2,2)
plt.title("Mirror Padding")
plt.imshow(mirror_pad, cmap='gray')
plt.axis('off')

print("Original Shape:", img.shape)
print("Mirror Padding Shape:", mirror_pad.shape)

plt.show()
```
